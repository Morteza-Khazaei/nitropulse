# Nitropulse Target Validation Plots

This notebook demonstrates how to use `nitropulse.utils.validation` to generate scatter and time-series validation graphics for any variable available in the holdout dataframe (e.g. soil surface moisture, soil temperature, etc.).

## 1. Imports
Import the plotting helpers and supporting libraries.

```python
import pandas as pd
from nitropulse.utils.validation import summarize_crop_station_years, summarize_crop_coverage
from nitropulse.utils.validation import (
    create_scatter_plot,
    create_timeseries_plot,
)
```

```python
risma_df = pd.read_csv('~/.nitropulse/outputs/risma_df.csv')
risma_df.head()
```

```python
model = 'AIEM'  # options: 'AIEM', 'PRISM1', 'SPM3D', 'I2EM'
holdout_df = pd.read_csv('/home/morteza/.nitropulse/outputs/aiem_test_df.csv')
holdout_df.head()
```

```python
holdout_df.lc.unique()
```

```python
coverage_report = summarize_crop_coverage(holdout_df)
coverage_report.head(25)
```

```python
station_year_report = summarize_crop_station_years(holdout_df)
station_year_report.head(30)
```

```python
# Example: view all years/crops for station MB24 on a single timeseries figure
figure_name = f'{model}_time_seiries_plot_cereals_fix_year_146'
fig = create_timeseries_plot(
    risma_df, holdout_df,
    station=None,
    year=None,
    crop=146,
    share_year=False, # all on one row
    precip_opacity=0.6,
    title_text=f'{model}: Cool-season cereals (146); similar BBCH pace & canopy form',
    # save_html=figure_name,
    save=figure_name
)
fig.show()
```

```python
stations = ['MB4', 'MB5', 'MB6', 'MB8', 'MB10', 'MB11', 'MB12', 'MB21', 'MB22', 'MB23', 'MB24', 'MB25']
figure_name = f'{model}_time_seiries_plot_cereals'

fig = create_timeseries_plot(
    risma_df, holdout_df,
    station=stations,
    year=None,
    crop=[133,136,146],
    share_year=True,
    precip_opacity=0.6,
    title_text=f'{model}: Cool-season cereals; similar BBCH pace & canopy form',
    # save_html=figure_name,
    save=figure_name
)
fig.show()
```

```python
stations = ['MB4', 'MB5', 'MB6', 'MB8', 'MB10', 'MB11', 'MB12', 'MB21', 'MB22', 'MB23', 'MB24', 'MB25']
figure_name = f'{model}_scatter_all_stations_cereals'

scatter_fig = create_scatter_plot(
    obs_df=risma_df,
    test_df=holdout_df,
    target='ssm',
    station=stations, # fix station
    year=None, # allow all available years
    crop=[133,136,146], # include all crop types
    title_text=f'{model}: Cool-season cereals; similar BBCH pace & canopy form',
    # save_html=figure_name,
    save=figure_name
)
scatter_fig.show()
```
